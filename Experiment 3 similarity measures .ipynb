## 3. Calculate similarity measures for a given dataset

```python
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity, euclidean_distances
from scipy.stats import pearsonr
```

```python
df = pd.read_csv("C:\\Users\HP-LSC-001\\Desktop\\Goura Nachika Recommend System LAB\\Experiment 3\\ratings.csv")   # <-- change the file name if needed

print("Raw Dataset:")
display(df)
```

```python
ratings_matrix = df.pivot_table(index='userId', 
                                columns='movieId', 
                                values='rating').fillna(0)

ratings_matrix
```

## Cosine Similarity (User–User)

```python
from sklearn.metrics.pairwise import cosine_similarity

cos_sim = pd.DataFrame(
    cosine_similarity(ratings_matrix),
    
    index=ratings_matrix.index,
    columns=ratings_matrix.index
)

print("Cosine Similarity Matrix:")
cos_sim
```

## Pearson Similarity

```python
pearson_sim = ratings_matrix.T.corr(method='pearson')

print("Pearson Similarity Matrix:")
pearson_sim
```

## Euclidean Similarity

```python
from sklearn.metrics import euclidean_distances

dist = pd.DataFrame(
    euclidean_distances(ratings_matrix),
    index=ratings_matrix.index,
    columns=ratings_matrix.index
)

euclidean_sim = 1 / (1 + dist)

print("Euclidean Similarity Matrix:")
euclidean_sim
```

## Jaccard Similarity (for implicit rating)

```python
binary_matrix = ratings_matrix.applymap(lambda x: 1 if x > 0 else 0)

jaccard_sim = pd.DataFrame(index=binary_matrix.index, columns=binary_matrix.index)

for u1 in binary_matrix.index:
    for u2 in binary_matrix.index:
        intersection = (binary_matrix.loc[u1] & binary_matrix.loc[u2]).sum()
        union = (binary_matrix.loc[u1] | binary_matrix.loc[u2]).sum()
        jaccard_sim.loc[u1, u2] = intersection / union

print("Jaccard Similarity Matrix:")
jaccard_sim
```
