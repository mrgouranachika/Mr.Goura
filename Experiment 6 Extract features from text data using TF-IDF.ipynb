# 6.Extract features from text data using TF-IDF

```python
# Import necessary libraries
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
# Sample text data
documents = [
    "The quick brown fox jumps over the lazy dog.",
    "Never jump over a lazy dog quickly.",
    "The dog is brown and lazy.",
    "A quick brown fox is a fast animal."
]
```

```python
# 1. Initialize TfidfVectorizer
# You can customize parameters like max_features, ngram_range, stop_words, etc.
# max_features: Limits the number of features (words) to consider.
# stop_words: Removes common words like "the", "a", "is", etc.
# ngram_range: Allows for considering sequences of words (e.g., bigrams).
tfidf_vectorizer = TfidfVectorizer(max_features=1000, stop_words='english')
```

```python
# 2. Fit and transform the documents
# This step learns the vocabulary and calculates TF-IDF scores for each document.
tfidf_matrix = tfidf_vectorizer.fit_transform(documents)
```

```python
# 3. Get feature names (words)
feature_names = tfidf_vectorizer.get_feature_names_out()
```

```python
# 4. Convert the TF-IDF matrix to a DataFrame for better readability (optional)
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=feature_names)
```

```python
# 5. Display the results
print("TF-IDF Matrix (Sparse Representation):")
print(tfidf_matrix)

print("\nTF-IDF Matrix (DataFrame Representation):")
print(tfidf_df)

print("\nFeature Names (Vocabulary):")
print(feature_names)

# You can also access the TF-IDF scores for a specific document or word
# For example, the TF-IDF scores for the first document:
print(f"\nTF-IDF scores for the first document:\n{tfidf_df.iloc[0]}")
```
